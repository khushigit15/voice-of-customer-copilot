```python
df_clean["is_source_artifact"] = (
    df_clean["feedback"]
    .str.lower()
    .str.strip()
    .isin([
        "translate review to english.",
        "report abuse. translate review to english",
        "."
    ])
)
```

```python
df_clean["is_source_artifact"].value_counts()
```

```python
low_info_patterns = [
    "ok",
    "ok. ok",
    "good",
    "good. good",
    "great",
    "great. great",
    "awesome",
    "excellent",
    "love it",
    "no",
    "no. no",
    "n/a",
    "n/a. n/a",
    "ugh",
    "ugh. ugh"
]
```

```python
df_clean["is_low_information"] = (
    df_clean["feedback"]
    .str.lower()
    .str.strip()
    .isin(low_info_patterns)
)
```

```python
df_clean["is_low_information"].value_counts()
```

```python
df_clean["is_usable_feedback"] = (
    ~df_clean["is_source_artifact"]
    & ~df_clean["is_low_information"]
    & (df_clean["feedback"].str.len() >= 10)
)
```

```python
df_clean["is_usable_feedback"].value_counts()
```

```python
df_clean.groupby("is_usable_feedback")["feedback"].count()
```

```python
df_clean.to_csv(
    "../data/processed/feedback_cleaned.csv",
    index=False
)
```

```python
df_clean["text_for_ai"] = (
    df_clean["review_title"].fillna("") + " " +
    df_clean["review_text"].fillna("")
).str.strip()
```

```python
rating_title_pattern = r"^(one|two|three|four|five)\s+stars?$"

df_clean["clean_title"] = (
    df_clean["review_title"]
    .str.strip()
)

df_clean.loc[
    df_clean["clean_title"].str.lower().str.match(
        rating_title_pattern,
        na=False
    ),
    "clean_title"
] = ""
```

```python
df_clean["text_for_ai"] = (
    df_clean["clean_title"] + " " +
    df_clean["review_text"]
).str.strip()
```

```python
df_clean.to_csv(
    "../data/processed/feedback_cleaned.csv",
    index=False
)
```
